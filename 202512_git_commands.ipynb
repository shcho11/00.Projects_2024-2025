```python
# git init #현재 디렉토리를 git 저장소로 초기화
# git status # 현재 디렉토리 상태를 한 눈에 점검 (commit,push 해도되는지)

# git remote rm origin
  # 로컬 저장소에 등록된 origin 원격저장소의 '연결 정보'(url 매핑정보)를 삭제 (upstream 연결도 해제)
# git remote add origin git@~~~.git
  # 원격저장소주소를 origin 으로 등록
# git push -u origin main
  # upstream 을 해제한 경우, main ->origin/main 으로 최초 업로드, 동시에 upstream 설정
  # 이걸 하지 않으면 어느 원격 브랜치와 연결할 지 안정해져있어서 git push 단독 실행으로는 원격 반영 X
  # 또는 git push --set-upstream origin main 도 가능

# git add . #현재 디렉토리의 모든 변경파일을 staging 영역에 등록
# git commit -m "최초 업로드" #로컬저장소에 첫 커밋 생성

# git branch -M main          # main 브랜치 명시 (현재 브랜치를 강제로 main 으로 이름변경)
# git push -u origin main     # 최초 푸시 + upstream 설정

import subprocess as sbp
import time

commands = [
    'git status',
    'git add',
    'git config user.email "abcedfg@hij.com"',
    'git config user.name "xyz"',
    'git branch -M main',
    'git commit -, "commit_name"',
    'git push',

]

for command in commands :
  print(command)

  try :
    sbp.run([command], shell=True, check=True)

  except :
    time.sleep(3)
```

```python
# 커밋, 푸시, 상태 확인 용도 코드들

# 최근 커밋 5개를 짧게 (해시 + 메시지) 확인 (내 커밋이 실제 로컬에 존재하는지)
!git log --oneline -n 5
# 위와 기능동일하지만, 브랜치 / 태그포인터까지 표시 (origin/main 라벨이 붙어있으면 원격 추적 브랜치가 로컬에 존재갱신됨 의미)
!git log --oneline --decorate n

# 최신 커밋(HEAD)의 상세 내용 + 변경된 파일 목록 확인 (무슨 파일이 커밋되었는지 확인용)
!git show --name-only HEAD  #

# 원격 저장소 목록과 URL 확인 (fetch / push)
!git remote -v

# 로컬 브랜치 목록 (각 브랜치의 최신 커밋 + upstream(추적대상) 표시)
!git branch -vv

# 원격에 실제로 존재하는 브랜치들을 서버에서 조회
!git ls-remote --heads origin

# push 과정을 자세한 로그로 보기
# 만약 upstream 없다면 "The current branch main has no upstream branch ..." 출력
!git push -v
```
