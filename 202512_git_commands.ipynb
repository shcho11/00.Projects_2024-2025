```python
# git init #현재 디렉토리를 git 저장소로 초기화
# git status

# git remote rm origin # 이미 origin 있었을 경우
# git remote add origin git@~~~.git #원격저장소주소를 origin 으로 등록

# git add . #현재 디렉토리의 모든 변경파일을 staging 영역에 등록
# git commit -m "최초 업로드" #로컬저장소에 첫 커밋 생성

# git branch -M main          # main 브랜치 명시 (현재 브랜치를 강제로 main 으로 이름변경)
# git push -u origin main     # 최초 푸시 + upstream 설정

import subprocess as sbp
import time

commands = [
    'git status',
    'git add',
    'git config user.email "abcedfg@hij.com"',
    'git config user.name "xyz"',
    'git branch -M main',
    'git commit -, "commit_name"',
    'git push',

]

for command in commands :
  print(command)

  try :
    sbp.run([command], shell=True, check=True)

  except :
    time.sleep(3)
```
