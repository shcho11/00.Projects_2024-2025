### Business Day 체크해서 자동화 돌리기 (files rm automation)

```python
import os
import time
from datetime import datetime
from pandas.tseries.offsets import CustomBusinessDay, BDay

os.environ['TZ'] = 'Asia/Seoul'
time.tzset()

class DateCheck :

  def __init__(self) :

    self.holidays = ['2026-01-01','2026-02-16','2026-02-17','2026-02-18',
                     '2026-03-02',
                     '2026-05-01','2026-05-05','2026-05-25','2026-06-03',
                     '2026-08-17','2026-09-24','2026-09-25',
                     '2026-10-05','2026-10-09','2026-12-25'] #korean national holidays

    self.holidays_s = [x.replace("-","") for x in self.holidays]

    self.today = datetime.today()
    self.cbd = CustomBusinessDay(holidays=self.holidays)
    self.priorday = self.today - self.cbd

    # strftime
    self.today_yymmdd = self.today.strftime("%Y%m%d")
    self.priorday_yymmdd = self.priorday.strftime("%Y%m%d")

    self.date_list = []

  def run_control(self) :

      if self.today.weekday() in (5,6) :
        print("Today is Weekend. Bye~!")
        time.sleep(1)
        os._exit(0)

      elif self.today_yymmdd in self.holidays_s :
        print("Today is National Holiday. Bye~!")
        time.sleep(1)
        os._exit(0)

      else :
        print("Date Check Complete")

  def rm_dates(self, start, end) :

    for i in range(end) :

      if i == 0 :
        date = self.priorday

      elif i < start :
        date = date - self.cbd

      else :
        date = date - self.cbd
        date_yymmdd = date.strftime("%Y%m%d")
        self.date_list.append(date_yymmdd)

    return self.date_list

DC = DateCheck()
DC.run_control()

list1 = DC.rm_dates(20,30)
print(list1)

# before rm
print("len listdir(folder1) before rm :", len(os.listdir('home/folder1')))
print("len listdir(folder2) before rm :", len(os.listdir('home/folder2')))
print("len listdir(folder3) before rm :", len(os.listdir('home/folder3')))

for strdate in list1 :
  !find ./folder1/ -type f -name *{strdate}*
  !find ./folder2/ -type f -name *{strdate}*
  !find ./folder3/ -type f -name *{strdate}*

# rm csv
for strdate in list1 :
  !rm ./folder1/*{strdate}*.csv
  !rm ./folder2/*{strdate}*.csv
  !rm ./folder3/*{strdate}*.csv

# rm xlsx
for strdate in list1 :
  !rm ./folder1/*{strdate}*.xlsx
  !rm ./folder2/*{strdate}*.xlsx
  !rm ./folder3/*{strdate}*.xlsx

# after rm
print("len listdir(folder1) after rm :", len(os.listdir('home/folder1')))
print("len listdir(folder2) after rm :", len(os.listdir('home/folder2')))
print("len listdir(folder3) after rm :", len(os.listdir('home/folder3')))

for strdate in list1 :
  !find ./folder1/ -type f -name *{strdate}*
  !find ./folder2/ -type f -name *{strdate}*
  !find ./folder3/ -type f -name *{strdate}*

```
